```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Webcam not detected!")
else:
    ret, frame = cap.read()

    if ret:
        plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        plt.title("Webcam Capture")
        plt.axis("off")
        plt.show()

cap.release()
```
